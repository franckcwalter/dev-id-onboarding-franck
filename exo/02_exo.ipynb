```python
import numpy as np

# Crée un tableau temperatures de 30 valeurs simulées (entre 15 et 30 °C)
# avec un random.default_rng(seed=42)
temperatures = np.random.default_rng(seed=42).uniform(low=15, high=30,size=30)
print(f"Liste aléatoire de températures : {temperatures}")

# Calcule la moyenne, la médiane et l'écart-type.
print(f"Moyenne des températures: {temperatures.mean().round(1)}")
print(f"Écart-type des températures: {temperatures.std().round(1)}")
print(f"Médiane des températures: {np.median(temperatures).round(1)}")
```

```python
## module 2 ?

# Combien de jours sont au-dessus de 25 °C ?

# hot_day_amount = len(temperatures[temperatures > 25])
hot_day_amount = int((temperatures > 25).sum())
print(f"Il y a {hot_day_amount} jours >25°C.")
```

```python
# Remplace toutes les valeurs au-dessus de 28 °C par np.nan
temperatures[temperatures > 28] = np.nan
print(temperatures)

# puis recalcule la moyenne en ignorant les NaN.
# new_mean = temperatures[~np.isnan(temperatures)].mean().round(1)
new_mean = np.nanmean(temperatures).round(1)

print(f"Nouvelle moyenne (sans les temp > 28°C devenues NaN) : {new_mean} °C")

## en créant un nouveau tableau
temperatures_without_high = np.where(temperatures > 28, np.nan, temperatures)
```
