## 5. Réponses aux 3 questions d'acculturation IA

### 5.1 Apprentissage supervisé vs non supervisé

Dans l'apprentissage supervisé, on entraîne l'IA en lui montrant des exemple de bonne réponse et de mauvaise réponse, pour une question donnée.
Par exemple montrer des photos de chat et lui dire "ça c'est un chat" et "ça c'est pas un chat" afin qu'elle sache reconnaitre un chat si on lui fournit une image.

Dans l'apprentissage non-supervisé, on laisse la machine trouver ce qui va ressortir comme caractéristiques distinctifs dans un ensemble de données.
Par exemple, trouver ce qui caractérise ma base de données clients.

### 5.2 À quoi sert le transfer learning ?

Le transfert learning permet de réutiliser un modèle existant puis de le réentraîner avec des données spécifiques, afin qu'il soit performant sur la tâche sur laquelle on l'implémente. Cela permet de ne pas partir de zéro et d'entraîner un modèle avec peu de données.

Exemple: utiliser un modèle de vision sur un jeu d'images particulières, comme pour la reconnaissance de maladies sur des radios peut-être.

### 5.3 Pourquoi un LLM n'est pas toujours la bonne réponse ?

Les LLM sont utlies pour les tâches qui impliquent des longs textes dont le format est différent. Le coût d'inférence est élevé et donc ils ne sont pas la meilleure solution pour des problèmes qui peuvent être résolus par des modèles plus réduits.

Par exemple, un LLM serait sur-dimensionné pour une chaine de production de boîtes de pétri qui veut reconnaître des défauts dans la production, dans le cas où elle dispose de photos labellisées comme boîtes avec des défaut et boîte sans défaut.
